# 📈 Customer Acquisition Trend Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

cs['SignupDate'] = pd.to_datetime(cs['SignupDate'])

monthly = (
    cs.groupby(cs['SignupDate'].dt.to_period('M'))
      .size()
      .reset_index(name='Customers')
)

monthly['SignupDate'] = monthly['SignupDate'].astype(str)

plt.figure(figsize=(14,7))
sns.lineplot(
    data=monthly,
    x='SignupDate',
    y='Customers',
    marker='o',
    linewidth=3,
    color='#00CFFF'
)

plt.xticks(rotation=45)
plt.title("Customer Acquisition Trend")
plt.xlabel("Month")
plt.ylabel("New Customers")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 🎯 Business Objective
This visualization tracks the number of new customers acquired each month, helping stakeholders evaluate growth momentum, customer acquisition effectiveness, and overall business expansion.

# 📈 Cumulative Customer Growth Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

cs['SignupDate'] = pd.to_datetime(cs['SignupDate'])

daily = (
    cs.groupby('SignupDate')
      .size()
      .reset_index(name='Customers')
)

daily['Cumulative'] = daily['Customers'].cumsum()

plt.figure(figsize=(14,7))

plt.plot(
    daily['SignupDate'],
    daily['Cumulative'],
    linewidth=4,
    color='#6C5CE7'
)

plt.fill_between(
    daily['SignupDate'],
    daily['Cumulative'],
    alpha=0.3,
    color='#6C5CE7'
)

plt.title("Cumulative Customer Growth")
plt.xlabel("Date")
plt.ylabel("Total Customers")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib as mpl
from matplotlib.patches import Circle

# ==========================================
# FONT SETTINGS
# ==========================================

mpl.rcParams['font.family'] = 'Segoe UI'
mpl.rcParams['axes.unicode_minus'] = False

# ==========================================
# LOAD DATA
# ==========================================

#cs = pd.read_csv("Customers.csv")

cs['SignupDate'] = pd.to_datetime(cs['SignupDate'])

cs['Year'] = cs['SignupDate'].dt.year
cs['Month'] = cs['SignupDate'].dt.month_name()
cs['Quarter'] = cs['SignupDate'].dt.to_period('Q')

# ==========================================
# THEME COLORS
# ==========================================

plt.style.use('dark_background')

BG = '#0B1220'
PANEL = '#131C2F'

BLUE = '#00D9FF'
PURPLE = '#9D4EDD'
ORANGE = '#FF9F1C'
GREEN = '#00E676'

# ==========================================
# KPI VALUES
# ==========================================

total_customers = len(cs)
total_regions = cs['Region'].nunique()

first_signup = cs['SignupDate'].min().strftime('%Y-%m-%d')
last_signup = cs['SignupDate'].max().strftime('%Y-%m-%d')

# ==========================================
# CREATE DASHBOARD
# ==========================================

fig = plt.figure(
    figsize=(24, 16),
    constrained_layout=True
)

fig.patch.set_facecolor(BG)

gs = fig.add_gridspec(
    3,
    3
)

# ==========================================
# WATERMARK
# ==========================================

fig.text(
    0.5,
    0.5,
    "E-COMMERCE",
    fontsize=150,
    color="white",
    alpha=0.03,
    ha="center",
    va="center",
    rotation=25,
    fontweight='bold'
)


# ==========================================
# 1 ACQUISITION TREND
# ==========================================

ax1 = fig.add_subplot(gs[0, :2])
ax1.set_facecolor(PANEL)

monthly = (
    cs.groupby(cs['SignupDate'].dt.to_period('M'))
      .size()
      .reset_index(name='Customers')
)

monthly['Month'] = monthly['SignupDate'].astype(str)

sns.lineplot(
    data=monthly,
    x='Month',
    y='Customers',
    marker='o',
    linewidth=4,
    color=BLUE,
    ax=ax1
)

ax1.set_title(
    'Customer Acquisition Trend',
    fontsize=18,
    fontweight='bold'
)

ax1.tick_params(
    axis='x',
    rotation=45,
    labelsize=8
)

ax1.grid(alpha=0.15)

# ==========================================
# 2 DONUT CHART
# ==========================================

ax2 = fig.add_subplot(gs[0, 2])
ax2.set_facecolor(PANEL)

region_counts = cs['Region'].value_counts()

colors = [BLUE, ORANGE, PURPLE, GREEN]

ax2.pie(
    region_counts,
    labels=region_counts.index,
    colors=colors,
    autopct='%1.1f%%',
    wedgeprops=dict(width=0.4)
)

centre_circle = Circle(
    (0, 0),
    0.60,
    fc=PANEL
)

ax2.add_artist(centre_circle)

ax2.text(
    0,
    0,
    "ECOM",
    fontsize=20,
    fontweight='bold',
    color=BLUE,
    ha='center'
)

ax2.set_title(
    'Regional Customer Share',
    fontsize=18,
    fontweight='bold'
)

# ==========================================
# 3 HEATMAP
# ==========================================

ax3 = fig.add_subplot(gs[1, 0])

month_order = [
    'January','February','March','April',
    'May','June','July','August',
    'September','October','November','December'
]

heatmap_data = (
    pd.crosstab(cs['Month'], cs['Year'])
      .reindex(month_order)
      .fillna(0)
)

sns.heatmap(
    heatmap_data,
    cmap='YlOrBr',
    annot=True,
    fmt='.0f',
    linewidths=0.5,
    ax=ax3
)

ax3.set_title(
    'Customer Signup Heatmap',
    fontsize=18,
    fontweight='bold'
)

# ==========================================
# 4 CUMULATIVE GROWTH
# ==========================================

ax4 = fig.add_subplot(gs[1, 1])
ax4.set_facecolor(PANEL)

daily = (
    cs.groupby('SignupDate')
      .size()
      .reset_index(name='Customers')
      .sort_values('SignupDate')
)

daily['Cumulative'] = daily['Customers'].cumsum()

ax4.plot(
    daily['SignupDate'],
    daily['Cumulative'],
    color=PURPLE,
    linewidth=4
)

ax4.fill_between(
    daily['SignupDate'],
    daily['Cumulative'],
    color=PURPLE,
    alpha=0.30
)

ax4.set_title(
    'Cumulative Customer Growth',
    fontsize=18,
    fontweight='bold'
)

ax4.grid(alpha=0.15)

# ==========================================
# 5 REGION-WISE TREND
# ==========================================

ax5 = fig.add_subplot(gs[1, 2])
ax5.set_facecolor(PANEL)

region_growth = (
    cs.groupby(
        [cs['SignupDate'].dt.to_period('M'),
         'Region']
    )
    .size()
    .reset_index(name='Customers')
)

region_growth['Month'] = (
    region_growth['SignupDate']
    .astype(str)
)

sns.lineplot(
    data=region_growth,
    x='Month',
    y='Customers',
    hue='Region',
    marker='o',
    linewidth=3,
    ax=ax5
)

ax5.tick_params(
    axis='x',
    rotation=45,
    labelsize=7
)

ax5.set_title(
    'Region-wise Growth',
    fontsize=18,
    fontweight='bold'
)

ax5.grid(alpha=0.15)

# ==========================================
# 6 QUARTERLY PERFORMANCE
# ==========================================

ax6 = fig.add_subplot(gs[2, :])
ax6.set_facecolor(PANEL)

quarterly = (
    cs.groupby('Quarter')
      .size()
      .reset_index(name='Customers')
)

quarterly['Quarter'] = quarterly['Quarter'].astype(str)

bars = ax6.bar(
    quarterly['Quarter'],
    quarterly['Customers'],
    color=ORANGE
)

for bar in bars:

    height = bar.get_height()

    ax6.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f'{int(height)}',
        ha='center',
        va='bottom',
        color='white',
        fontsize=11,
        fontweight='bold'
    )

ax6.set_title(
    'Quarterly Acquisition Performance',
    fontsize=20,
    fontweight='bold'
)

ax6.grid(
    axis='y',
    alpha=0.15
)

# ==========================================
# TITLES
# ==========================================



fig.text(
    0.5,
    0.94,
    'E-Commerce Intelligence Platform',
    ha='center',
    color='lightgray',
    fontsize=14
)

# ==========================================
# SHOW DASHBOARD
# ==========================================

plt.show()
```
